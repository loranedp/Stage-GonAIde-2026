```python
from PIL import Image
import torch

# Importation image
image_pil=Image.open("crop/img_2.jpg")
image_pil
```

```python
import torchvision.transforms as T 
transform=T.Compose([T.ToTensor(),T.Resize((360,360))])
image=transform(image_pil)[0:3,:,:]
```

```python
import matplotlib.pyplot as plt

rot=T.Compose([T.RandomRotation(degrees=20)])
image_rot=rot(image)

plt.figure(figsize=(5,5))
plt.subplot(1,2,1)
plt.imshow(image.permute(1,2,0))
plt.axis("off")
plt.subplot(1,2,2)
plt.imshow(image_rot.permute(1,2,0))
plt.axis("off")
plt.show()
```

```python
bright=T.Compose([T.ColorJitter(brightness=0.4)])
image_bright=bright(image)
contr=T.Compose([T.ColorJitter(contrast=0.4)])
image_contr=contr(image)

plt.figure(figsize=(8,5))
plt.subplot(1,3,1)
plt.imshow(image.permute(1,2,0))
plt.axis("off")
plt.subplot(1,3,2)
plt.imshow(image_bright.permute(1,2,0))
plt.axis("off")
plt.subplot(1,3,3)
plt.imshow(image_contr.permute(1,2,0))
plt.axis("off")
plt.show()
```

```python
scale=T.Compose([T.RandomAffine(degrees=0,scale=(0.8,1.2))])
image_scale=scale(image)
translate = T.Compose([T.RandomAffine(degrees=0, translate=(0.1, 0.1))])
image_translate=translate(image)

plt.figure(figsize=(8,5))
plt.subplot(1,3,1)
plt.imshow(image.permute(1,2,0))
plt.axis("off")
plt.subplot(1,3,2)
plt.imshow(image_scale.permute(1,2,0))
plt.axis("off")
plt.subplot(1,3,3)
plt.imshow(image_translate.permute(1,2,0))
plt.axis("off")
plt.show()
```
